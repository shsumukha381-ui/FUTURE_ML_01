```python
# ⚠️ EDIT THESE to choose your scope
FILTER_STORE = None     # e.g. 44   |  None = all stores
FILTER_FAMILY = None    # e.g. "GROCERY I"   |  None = all families

df = train_raw.copy()
df["date"] = pd.to_datetime(df["date"], errors="coerce")

if FILTER_STORE is not None:
    df = df[df["store_nbr"] == FILTER_STORE]
if FILTER_FAMILY is not None:
    df = df[df["family"] == FILTER_FAMILY]

scope_label = "All Stores, All Families (Total Company Sales)"
if FILTER_STORE is not None or FILTER_FAMILY is not None:
    scope_label = f"Store: {FILTER_STORE if FILTER_STORE is not None else 'All'} | Family: {FILTER_FAMILY if FILTER_FAMILY is not None else 'All'}"

print(f"Forecasting scope: {scope_label}")
print(f"Rows in scope: {len(df)}")
```

```python
daily = df.groupby("date", as_index=False).agg(sales=("sales", "sum"), onpromotion=("onpromotion", "sum"))
daily = daily.sort_values("date").reset_index(drop=True)

# Fill missing calendar days
full_range = pd.date_range(daily["date"].min(), daily["date"].max(), freq="D")
daily = daily.set_index("date").reindex(full_range)
daily.index.name = "date"
missing_before = daily["sales"].isna().sum()
daily["sales"] = daily["sales"].interpolate(method="linear").fillna(method="bfill").fillna(method="ffill")
daily["onpromotion"] = daily["onpromotion"].fillna(0)
daily = daily.reset_index()

# Sales can't be negative
daily["sales"] = daily["sales"].clip(lower=0)

print(f"Date range: {daily['date'].min().date()} → {daily['date'].max().date()}")
print(f"Total days: {len(daily)}  |  Missing days filled: {missing_before}")
daily.head()
```

```python
daily["is_holiday"] = 0

if holidays_raw is not None:
    hol = holidays_raw.copy()
    hol["date"] = pd.to_datetime(hol["date"], errors="coerce")
    # National holidays that were NOT transferred to another day are the ones that actually affect that date
    national_holidays = hol[(hol["locale"] == "National") & (hol["transferred"] == False)]["date"].unique()
    daily["is_holiday"] = daily["date"].isin(national_holidays).astype(int)
    print(f"Merged holidays: {daily['is_holiday'].sum()} holiday-days flagged")
else:
    print("holidays_events.csv not provided — skipping holiday feature (is_holiday will be 0 for all days)")

if oil_raw is not None:
    oil = oil_raw.copy()
    oil["date"] = pd.to_datetime(oil["date"], errors="coerce")
    oil = oil.rename(columns={"dcoilwtico": "oil_price"})
    daily = daily.merge(oil[["date", "oil_price"]], on="date", how="left")
    daily["oil_price"] = daily["oil_price"].fillna(method="ffill").fillna(method="bfill")
    print("Merged oil price")
else:
    daily["oil_price"] = 0
    print("oil.csv not provided — skipping oil price feature")

daily.head()
```

```python
fig, ax = plt.subplots()
ax.plot(daily["date"], daily["sales"], color="#2563eb", linewidth=1)
ax.set_title(f"Historical Sales Over Time — {scope_label}", fontsize=13, fontweight="bold")
ax.set_xlabel("Date")
ax.set_ylabel("Sales")
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))
fig.autofmt_xdate()
plt.tight_layout()
plt.show()
```

```python
data = daily.copy()

# Calendar features
data["year"] = data["date"].dt.year
data["month"] = data["date"].dt.month
data["day"] = data["date"].dt.day
data["day_of_week"] = data["date"].dt.dayofweek
data["day_of_year"] = data["date"].dt.dayofyear
data["week_of_year"] = data["date"].dt.isocalendar().week.astype(int)
data["is_weekend"] = (data["day_of_week"] >= 5).astype(int)

# Cyclical encodings
data["month_sin"] = np.sin(2 * np.pi * data["month"] / 12)
data["month_cos"] = np.cos(2 * np.pi * data["month"] / 12)
data["dow_sin"] = np.sin(2 * np.pi * data["day_of_week"] / 7)
data["dow_cos"] = np.cos(2 * np.pi * data["day_of_week"] / 7)

# Lag features
for lag in [1, 7, 14, 30]:
    data[f"lag_{lag}"] = data["sales"].shift(lag)

# Rolling averages (recent trend/momentum)
for window in [7, 14, 30]:
    data[f"rolling_mean_{window}"] = data["sales"].shift(1).rolling(window).mean()

data = data.dropna().reset_index(drop=True)

feature_cols = [c for c in data.columns if c not in ["date", "sales"]]
print(f"Number of features: {len(feature_cols)}")
data.head()
```

```python
test_size_days = 60

train = data.iloc[:-test_size_days]
test = data.iloc[-test_size_days:]

X_train, y_train = train[feature_cols], train["sales"]
X_test, y_test = test[feature_cols], test["sales"]

print(f"Train period: {train['date'].min().date()} → {train['date'].max().date()} ({len(train)} days)")
print(f"Test period:  {test['date'].min().date()} → {test['date'].max().date()} ({len(test)} days)")
```

```python
models = {
    "Linear Regression": LinearRegression(),
    "Random Forest": RandomForestRegressor(n_estimators=300, max_depth=10, random_state=42),
}

predictions = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    predictions[name] = model.predict(X_test)

predictions["Baseline (naive, last week)"] = test["lag_7"].values

print("Models trained ✅")
```

```python
def mape(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    mask = y_true != 0
    return np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100

results = []
for name, preds in predictions.items():
    mae = mean_absolute_error(y_test, preds)
    rmse = np.sqrt(mean_squared_error(y_test, preds))
    mape_val = mape(y_test, preds)
    results.append({"Model": name, "MAE": round(mae, 2), "RMSE": round(rmse, 2), "MAPE (%)": round(mape_val, 2)})

results_df = pd.DataFrame(results).sort_values("MAPE (%)")
results_df
```

```python
fig, ax = plt.subplots()
ax.plot(test["date"], y_test, label="Actual Sales", color="black", linewidth=2)

colors = {"Linear Regression": "#2563eb", "Random Forest": "#16a34a", "Baseline (naive, last week)": "#9ca3af"}
for name, preds in predictions.items():
    style = "--" if "Baseline" in name else "-"
    ax.plot(test["date"], preds, label=name, linestyle=style, color=colors.get(name), linewidth=1.8)

ax.set_title(f"Forecast Accuracy Check — {scope_label}", fontsize=13, fontweight="bold")
ax.set_xlabel("Date")
ax.set_ylabel("Sales")
ax.legend()
fig.autofmt_xdate()
plt.tight_layout()
plt.show()
```

```python
best_model_name = results_df.iloc[0]["Model"]
print(f"Best model: {best_model_name}")

if best_model_name == "Baseline (naive, last week)":
    best_model = models["Linear Regression"]
    best_model.fit(X_train, y_train)
else:
    best_model = models[best_model_name]
    best_model.fit(data[feature_cols], data["sales"])

FORECAST_DAYS = 30
last_oil_price = data["oil_price"].iloc[-1]
future_holidays = []  # ⚠️ optionally add known future holiday dates here, e.g. pd.Timestamp("2017-12-25")

history = data.copy()
future_rows = []
last_date = history["date"].max()

for i in range(1, FORECAST_DAYS + 1):
    next_date = last_date + pd.Timedelta(days=i)
    row = {"date": next_date}
    row["year"] = next_date.year
    row["month"] = next_date.month
    row["day"] = next_date.day
    row["day_of_week"] = next_date.dayofweek
    row["day_of_year"] = next_date.dayofyear
    row["week_of_year"] = next_date.isocalendar()[1]
    row["is_weekend"] = int(next_date.dayofweek >= 5)
    row["month_sin"] = np.sin(2 * np.pi * row["month"] / 12)
    row["month_cos"] = np.cos(2 * np.pi * row["month"] / 12)
    row["dow_sin"] = np.sin(2 * np.pi * row["day_of_week"] / 7)
    row["dow_cos"] = np.cos(2 * np.pi * row["day_of_week"] / 7)
    row["is_holiday"] = int(next_date in future_holidays)
    row["oil_price"] = last_oil_price
    row["onpromotion"] = data["onpromotion"].iloc[-30:].mean()  # assume similar promo activity to recent average

    combined = pd.concat(
        [history[["date", "sales"]],
         pd.DataFrame(future_rows)[["date", "sales"]] if future_rows else pd.DataFrame(columns=["date", "sales"])],
        ignore_index=True
    )

    for lag in [1, 7, 14, 30]:
        lag_date = next_date - pd.Timedelta(days=lag)
        match = combined.loc[combined["date"] == lag_date, "sales"]
        row[f"lag_{lag}"] = match.values[0] if len(match) else combined["sales"].iloc[-1]

    for window in [7, 14, 30]:
        recent = combined["sales"].iloc[-window:]
        row[f"rolling_mean_{window}"] = recent.mean()

    X_next = pd.DataFrame([row])[feature_cols]
    pred_sales = best_model.predict(X_next)[0]
    row["sales"] = pred_sales
    future_rows.append(row)

future_df = pd.DataFrame(future_rows)[["date", "sales"]]
future_df["sales"] = future_df["sales"].clip(lower=0)
future_df.columns = ["date", "forecast"]
future_df
```

```python
fig, ax = plt.subplots()

recent_history = daily[daily["date"] >= daily["date"].max() - pd.Timedelta(days=120)]
ax.plot(recent_history["date"], recent_history["sales"], label="Historical Sales", color="#1f2937", linewidth=1.8)
ax.plot(future_df["date"], future_df["forecast"], label=f"Forecast (next {FORECAST_DAYS} days)",
        color="#dc2626", linewidth=2, linestyle="--", marker="o", markersize=3)

ax.axvline(daily["date"].max(), color="gray", linestyle=":", linewidth=1)
ax.text(daily["date"].max(), ax.get_ylim()[1]*0.95, "  Last known date", color="gray")

ax.set_title(f"Sales Forecast — Next {FORECAST_DAYS} Days ({scope_label})", fontsize=13, fontweight="bold")
ax.set_xlabel("Date")
ax.set_ylabel("Sales")
ax.legend()
fig.autofmt_xdate()
plt.tight_layout()
plt.show()

print(f"Average predicted daily sales over next {FORECAST_DAYS} days: {future_df['forecast'].mean():.1f}")
print(f"Total predicted sales over next {FORECAST_DAYS} days: {future_df['forecast'].sum():.1f}")
```
